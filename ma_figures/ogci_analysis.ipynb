```python
pdata = pd.read_excel('comparison_table.xlsx', index_col=[1, 0], usecols=list(range(5)))
pdata.columns = pdata.columns.str.rstrip(' (Gt CO2)')
pdata
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 6))

ax = axs[0]
pdata.loc['Discovered'].plot.bar(logy=True, cmap='viridis', ax=ax)
ax.set_ylabel('CO$_{2}$ storage potential (Gt)')
ax.set_xlabel('')
ax.set_ylim(0, 1e4)
ax.set_title("Discovered vs.\n Prudent Assessments")

ax = axs[1]
pdata.loc['Undiscovered + Discovered'].plot.bar(logy=True, cmap='viridis', ax=ax, legend=False)
#ax.set_ylabel('CO$_{2}$ storage potential (Gt)')
ax.set_xlabel('')
ax.set_ylim(0, 1e4)
ax.set_title("Discovered and Undiscovered vs.\n Technical Potential Assessments")
```

```python
fig, axs = plt.subplots(1, 2, figsize=(10, 6))

ax = axs[0]
pdata.loc['Discovered'].plot.bar(logy=False, cmap='viridis', ax=ax)
ax.set_ylabel('CO$_{2}$ storage potential (Gt)')
ax.set_xlabel('')
#ax.set_ylim(0, 1e4)
ax.set_title("Discovered vs.\n Prudent Assessments")

ax = axs[1]
pdata.loc['Undiscovered + Discovered'].plot.bar(logy=False, cmap='viridis', ax=ax, legend=False)
#ax.set_ylabel('CO$_{2}$ storage potential (Gt)')
ax.set_xlabel('')
#ax.set_ylim(0, 1e4)
ax.set_title("Discovered and Undiscovered vs.\n Technical Potential Assessments")
```
