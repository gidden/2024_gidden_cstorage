```python
pdata = pd.read_excel('comparison_table.xlsx', index_col=[0, 1], usecols=list(range(5)))
pdata.columns = pdata.columns.str.rstrip(' (Gt CO2)')
pdata
```

```python
x = pdata.reset_index()
x = x.assign(label=x['Country'] + ' ' + x['Type']).set_index('label').drop(['Country', 'Type'], axis=1)
ax = x.plot.bar(logy=False, cmap='viridis')
ax.set_ylabel('CO$_{2}$ storage potential (Gt)', fontsize=14)
ax.set_xlabel('')
ax = x.plot.bar(logy=True, cmap='viridis')
ax.set_ylabel('CO$_{2}$ storage potential (Gt)', fontsize=14)
ax.set_xlabel('')
```
