```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
mincap = {'IPCC 2005': 1000,
          'Kearns et al., 2017': 8000,
          'Wei et al., 2024': np.nan,
          'OGCI, 2025': 677,
          'Gidden et al., 2025': 1285}
maxcap = {'IPCC 2005': 10000,
          'Kearns et al., 2017': 55000,
          'Wei et al., 2024': np.nan,
          'OGCI, 2025': 14256,
          'Gidden et al., 2025': 11800}
gidden_max = 2710
gidden_med = 1450
```

```python
fig, axs = plt.subplots(1, figsize=[6, 6])
axs.grid(which='major', axis='x')
for i in range(len(mincap)):
    axs.plot([i, i], [list(mincap.values())[i], list(maxcap.values())[i]], 'k', alpha=0.5)

axs.plot(range(len(mincap)), mincap.values(), 'ok', c='#fc9272', label='Assessed Minimum')
axs.plot(range(len(maxcap)), maxcap.values(), 'ob', c='#de2d26', label='Assessed Maximum')
x = np.array([4-.1, 4+.1])
axs.fill_between(x, 1285, gidden_max, color='#756bb1', label='Prudent Range')

axs.plot(2, 2082, marker='*', markersize=12, markerfacecolor='w', mec='k')
axs.set_xticks(range(len(maxcap)), list(maxcap.keys()), rotation=90, fontsize=14)
axs.set_ylabel('CO$_{2}$ storage potential (Gt)', fontsize=14)
axs.legend(fontsize=14)
fig.savefig('context.png', bbox_inches='tight')
```
